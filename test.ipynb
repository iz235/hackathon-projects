```python
%pip install matplotlib numpy
import matplotlib.pyplot as plt
import numpy as np

# Configuration du style
plt.style.use('ggplot')
colors = ['#2C3E50', '#3498DB', '#16A085', '#E74C3C']

# --- GRAPHIQUE 1 : PERCEPTION DE L'ÂGISME EN FRANCE ---
def plot_stats_france():
    categories = ['Victimes de discrim.', 'Témoins de stéréotypes', 'Crainte du vieillissement', 'Impact Emploi (+50 ans)']
    valeurs = [38, 72, 54, 65]  # Chiffres hypothétiques réalistes pour l'exemple

    plt.figure(figsize=(10, 6))
    bars = plt.bar(categories, valeurs, color=colors[1])
    plt.title('L\'Âgisme en France : État des lieux (%)', fontsize=14)
    plt.ylabel('Pourcentage de la population sondée')
    plt.ylim(0, 100)
    
    for bar in bars:
        height = bar.get_height()
        plt.text(bar.get_x() + bar.get_width()/2., height,
                 f'{height}%', ha='center', va='bottom')
    
    plt.tight_layout()
    plt.savefig('stats_france.png', dpi=300)
    print("Généré : stats_france.png")

# --- GRAPHIQUE 2 : ANALYSE CONCURRENTIELLE (RADAR) ---
def plot_concurrence():
    labels = np.array(['Innovation IA', 'Accessibilité', 'Pédagogie', 'Prix', 'Support Fr'])
    # Notre solution
    stats_nous = np.array([9, 8, 10, 9, 10])
    # Concurrence France (ex: Formation classique)
    stats_fr = np.array([2, 6, 8, 4, 9])
    # Concurrence Inter (ex: Apps US)
    stats_int = np.array([8, 7, 5, 6, 2])

    angles = np.linspace(0, 2*np.pi, len(labels), endpoint=False).tolist()
    stats_nous = np.concatenate((stats_nous,[stats_nous[0]]))
    stats_fr = np.concatenate((stats_fr,[stats_fr[0]]))
    stats_int = np.concatenate((stats_int,[stats_int[0]]))
    angles += angles[:1]

    fig, ax = plt.subplots(figsize=(8, 8), subplot_kw=dict(polar=True))
    
    ax.plot(angles, stats_nous, 'o-', linewidth=2, label="Notre Projet", color=colors[2])
    ax.fill(angles, stats_nous, alpha=0.25, color=colors[2])
    
    ax.plot(angles, stats_fr, 'o-', linewidth=2, label="Concurrents FR (Classique)", color=colors[0])
    ax.plot(angles, stats_int, 'o-', linewidth=2, label="Apps Internationales", color=colors[3])

    ax.set_thetagrids(np.degrees(angles[:-1]), labels)
    plt.title('Comparatif Concurrentiel', y=1.1)
    plt.legend(loc='upper right', bbox_to_anchor=(1.1, 1.1))
    
    plt.tight_layout()
    plt.savefig('concurrence.png', dpi=300)
    print("Généré : concurrence.png")

# --- GRAPHIQUE 3 : PLAN FINANCIER SUR 12 MOIS ---
def plot_finance():
    mois = np.arange(1, 13)
    # Dépenses (Investissement initial fort puis maintenance)
    depenses = [5000, 3000, 2000, 1500, 1500, 1500, 1500, 1800, 1500, 1500, 2000, 1500]
    # Revenus (Croissance progressive)
    revenus = [0, 0, 500, 1000, 1800, 2500, 3500, 4500, 5500, 7000, 8500, 10000]
    
    # Cumul pour voir le point de rentabilité
    cumul_depenses = np.cumsum(depenses)
    cumul_revenus = np.cumsum(revenus)

    plt.figure(figsize=(10, 6))
    plt.plot(mois, cumul_depenses, label='Dépenses Cumulées', color=colors[3], linestyle='--', linewidth=2)
    plt.plot(mois, cumul_revenus, label='Revenus Cumulés', color=colors[2], linewidth=2.5)
    plt.bar(mois, revenus, alpha=0.3, label='Revenus Mensuels', color=colors[1])
    
    plt.title('Projection Financière & ROI (12 Mois)', fontsize=14)
    plt.xlabel('Mois')
    plt.ylabel('Euros (€)')
    plt.legend()
    plt.grid(True, linestyle='--', alpha=0.7)
    
    plt.tight_layout()
    plt.savefig('finance.png', dpi=300)
    print("Généré : finance.png")

if __name__ == "__main__":
    plot_stats_france()
    plot_concurrence()
    plot_finance()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Style moderne
plt.style.use('seaborn-v0_8-whitegrid')
colors = ['#FF6B6B', '#4ECDC4', '#556270', '#C7F464']

# --- GRAPHIQUE 1 : Marché Serious Game (Croissance) ---
def plot_market_growth():
    years = [2024, 2028, 2035]
    market_size = [10.5, 18.2, 35.9] # Milliards USD (Données fournies)
    
    plt.figure(figsize=(10, 6))
    bars = plt.bar(years, market_size, color=colors[1], width=2)
    
    plt.title('Croissance Mondiale du Marché "Serious Game"', fontsize=14, fontweight='bold')
    plt.ylabel('Milliards USD ($)')
    plt.xticks(years)
    plt.grid(axis='y', linestyle='--', alpha=0.7)
    
    # Ajout des valeurs
    for bar in bars:
        height = bar.get_height()
        plt.text(bar.get_x() + bar.get_width()/2., height + 0.5,
                 f'{height} Mrd $', ha='center', va='bottom', fontsize=12, fontweight='bold')
                 
    plt.tight_layout()
    plt.savefig('market_growth.png', dpi=300)
    print("Généré : market_growth.png")

# --- GRAPHIQUE 2 : Structure de Financement (Camembert) ---
def plot_funding_structure():
    labels = ['Subventions Publiques (CNC/FAJV)', 'Sponsoring Privé (Mutuelles)', 'Apport Propre', 'Revenus (Licences)']
    sizes = [50, 30, 10, 10]
    explode = (0.1, 0, 0, 0)  # Mettre en avant les subventions

    plt.figure(figsize=(8, 8))
    plt.pie(sizes, explode=explode, labels=labels, colors=colors, autopct='%1.1f%%',
            shadow=True, startangle=140, textprops={'fontsize': 12})
    plt.title('Stratégie de Financement Hybride', fontsize=14, fontweight='bold')
    
    plt.tight_layout()
    plt.savefig('funding_pie.png', dpi=300)
    print("Généré : funding_pie.png")

# --- GRAPHIQUE 3 : Prévisionnel Trésorerie (12 mois) ---
def plot_cashflow():
    months = np.arange(1, 13)
    # Dépenses (Dév lourd au début)
    expenses = [15000, 12000, 10000, 8000, 5000, 5000, 5000, 5000, 5000, 5000, 5000, 5000]
    # Revenus (Subventions au début, puis Licences)
    revenues = [20000, 0, 0, 15000, 2000, 3000, 4000, 5000, 6000, 8000, 10000, 12000]
    
    cumul_cash = np.cumsum(np.array(revenues) - np.array(expenses))

    plt.figure(figsize=(10, 6))
    plt.plot(months, cumul_cash, marker='o', linestyle='-', color=colors[2], linewidth=3, label='Trésorerie Nette')
    plt.bar(months, revenues, alpha=0.3, color=colors[3], label='Entrées (Subv + Ventes)')
    plt.bar(months, expenses, alpha=0.3, color=colors[0], label='Sorties (Dév + Salaires)')
    
    plt.axhline(0, color='black', linewidth=1, linestyle='--')
    plt.title('Plan de Trésorerie Prévisionnel (Année 1)', fontsize=14, fontweight='bold')
    plt.xlabel('Mois')
    plt.ylabel('Euros (€)')
    plt.legend()
    plt.grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.savefig('cashflow.png', dpi=300)
    print("Généré : cashflow.png")

if __name__ == "__main__":
    plot_market_growth()
    plot_funding_structure()
    plot_cashflow()
```
